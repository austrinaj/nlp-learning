```python
import sys
sys.path.append("../week2")


from text_analyzer import tokenize, load_stopwords, remove_stopwords

stopwords = load_stopwords("../week2/stopwords.txt")
```

```python
df["tokens"] = df["text"].apply(tokenize)
df
```

```python
df["word_count"] = df["tokens"].apply(len)
df
```

```python
df["positive"] = df["rating"] >= 4
df
```

```python
df.groupby("positive")["word_count"].mean()
```

```python
good = df[df["positive"] == True]
good.shape
```

```python
bad = df[df["positive"] == False]
bad.shape
```

```python
from collections import Counter


good_words = []
for tokens in good["tokens"]:
    good_words.extend(tokens)
    good_words = remove_stopwords(good_words, stopwords)
Counter(good_words).most_common(5)

```

```python
def top_words(table, n=5):
        """Return the n most common words in the table's tokens, excluding stopwords."""
        words = []
        for tokens in table["tokens"]:
            words.extend(tokens)

        filtered = remove_stopwords(words, stopwords)
        return Counter(filtered).most_common(n)
```

```python
top_words(bad)
```
