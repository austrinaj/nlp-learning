```python
import numpy as np


prices_list = [100, 200, 300]
prices_array = np.array([100, 200, 300])


print(prices_list * 2)
print(prices_array * 2)
```

```python
counts = np.array([1, 0, 2]) #how many times each word appears
weights = np.array([0.5, -1.2, 0.3]) #weight of each word


print(counts * weights)
print((counts * weights).sum())
```

```python
intercept = -1.5
score = counts @ weights + intercept
print(score)
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
print(sigmoid(1.1))
print(sigmoid(-0.4))
print(sigmoid(0))
```

```python
x = np.array([[1, 0, 2], [0, 3, 0], [2, 1, 1]])
weights = np.array([0.5, -1.2, 0.3])
print(x.shape)
print(x @ weights)
```
